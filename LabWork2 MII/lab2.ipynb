### 6. Определение проблем данных
В любых данных могут быть «болячки». В наших датасетах есть две главные проблемы:
1. **Выбросы (аномалии):** Это числа, которые сильно отличаются от остальных. Например, если у всех программистов зарплата 100 тысяч, а у директора — 5 миллионов, это выброс. В финансах (золото и кофе) выбросы бывают из-за кризисов или паники на бирже, когда цена резко падает или взлетает.
2. **Пропуски:** Это пустые клетки в таблице (NaN). Они могут появляться, если биржа не работала в выходной день или если человек не указал свой опыт в анкете.

```python
print("=== ПРОВЕРКА НА ПРОБЛЕМЫ ===")

# 1. Проверяем пропуски (пустые клетки)
gold_missing = gold_df.isnull().sum().sum()
sal_missing = salaries_df.isnull().sum().sum()
print(f"1. Пропуски (пустые клетки) в золоте: {gold_missing} шт.")
print(f"   Пропуски в зарплатах: {sal_missing} шт.")

# 2. Доказываем наличие выбросов в зарплатах
print("\n2. Доказательство выбросов в зарплатах:")
max_sal = salaries_df['salary_in_usd'].max()
mean_sal = salaries_df['salary_in_usd'].mean()
print(f"   Средняя зарплата: {mean_sal:.0f} $")
print(f"   Самая высокая зарплата (выброс): {max_sal} $")
print("   (Видно, что максимум сильно больше среднего — это и есть выброс)")
```

### 7. Решение обнаруженных проблем
Мы не будем тупо удалять всё подряд, а подойдем с умом:
*   **Выбросы в финансах:** Мы их **оставляем**. Если в 2008 году цена золота рухнула, это не ошибка, а исторический факт. Удалять это нельзя, иначе модель не узнает, что кризисы бывают.
*   **Выбросы в зарплатах:** А вот в зарплатах огромные суммы директоров нам будут мешать. Мы **отрежем верхний 1%** (самые большие числа), чтобы модель не сбивала настройки на миллионерах.
*   **Пропуски:** Мы их обработаем в 10-м пункте (удалим или заполним).

```python
print("=== РЕШЕНИЕ ПРОБЛЕМЫ ВЫБРОСОВ В ЗАРПЛАТАХ ===")

# Находим 99-й процентиль (отсекаем самые extreme значения)
q_99 = salaries_df['salary_in_usd'].quantile(0.99)
print(f"Порог отсечения (99-й процентиль): {q_99:.0f} $")

# Оставляем только те строки, где зарплата меньше или равна порогу
salaries_filtered = salaries_df[salaries_df['salary_in_usd'] <= q_99]

print("\n--- РЕЗУЛЬТАТ ФИЛЬТРАЦИИ ---")
print(f"Было строк ДО фильтрации выбросов:    {len(salaries_df)}")
print(f"Стало строк ПОСЛЕ фильтрации (без 1%): {len(salaries_filtered)}")
print("Мы успешно убрали аномалии и сохранили остальные данные!")
```

### 9. Разбиение данных на выборки
Нельзя учить модель на всех данных сразу — она просто «зазубрит» ответы и провалится на реальном экзамене. Поэтому мы делим данные на три части:
1. **Обучающая (Train) — 60%:** Учебник. По нему модель учится.
2. **Контрольная (Validation) — 20%:** Пробный экзамен. На нем мы проверяем настройки модели.
3. **Тестовая (Test) — 20%:** Выпускной экзамен. Модель его никогда не видела, и мы честно проверяем, чему она научилась.

```python
from sklearn.model_selection import train_test_split

print("=== РАЗБИЕНИЕ ДАННЫХ НА ВЫБОРКИ ===")

# Сначала отделяем 40% данных для валидации и теста
train_data, temp_data = train_test_split(gold_df, test_size=0.4, random_state=42)
# Затем делим эти 40% пополам (по 20%)
val_data, test_data = train_test_split(temp_data, test_size=0.5, random_state=42)

print(f"1. Обучающая выборка (Train, 60%):     {len(train_data)} строк")
print(f"2. Контрольная выборка (Validation, 20%): {len(val_data)} строк")
print(f"3. Тестовая выборка (Test, 20%):          {len(test_data)} строк")
print(f"ИТОГО: {len(train_data) + len(val_data) + len(test_data)} строк (все данные на месте)")
```

### 11. Оценка сбалансированности выборок
Наши данные — это числа (цены). Но чтобы проверить, нет ли перекоса, я превратила задачу в «угадай, вырастет цена или упадет».
Я создала новый столбец `Target`: 
* Если завтрашняя цена больше сегодняшней — ставим `1` (цена выросла).
* Если меньше — ставим `0` (цена упала).
Затем я посчитала, сколько дней было с ростом, а сколько с падением. Если одних дней намного больше, чем других, это называется **дисбаланс**.

```python
print("=== ОЦЕНКА СБАЛАНСИРОВАННОСТИ ===")

# Создаем копию данных и новый столбец Target
gold_class = train_data.copy()
gold_class['Target'] = (gold_class['Close'].shift(-1) > gold_class['Close']).astype(int)
gold_class = gold_class.dropna(subset=['Target']) # Удаляем последнюю строку (у нее нет "завтра")

# Считаем, сколько раз цена росла, а сколько падала
counts = gold_class['Target'].value_counts()

print("Сколько дней было с падением цены (0):", counts.get(0, 0))
print("Сколько дней было с ростом цены (1): ", counts.get(1, 0))
print("\n(Если числа сильно отличаются, значит есть дисбаланс)")
```

### 12. Приращение данных (SMOTE и Undersampling)
Если в данных перекос (например, дней падения 500, а дней роста всего 300), модель может просто всегда отвечать «упадет» и будет права в 60% случаев. Чтобы её проучить, мы балансируем данные:
* **Oversampling (SMOTE):** Берет те 300 дней роста и создает на их основе искусственные копии, пока их не станет 500. Теперь у нас 500 на 500.
* **Undersampling:** Наоборот, берет 500 дней падения и удаляет оттуда лишние 200, чтобы осталось 300. Тоже получается 300 на 300.
Я применила оба метода из библиотеки `imbalanced-learn`.

```python
from imblearn.over_sampling import SMOTE
from imblearn.under_sampling import RandomUnderSampler

print("=== ПРИРАЩЕНИЕ ДАННЫХ ===")

# Отделяем признаки (X) от цели (y)
X = gold_class.drop(columns=['Target', 'Date'])
y = gold_class['Target']

# 1. Oversampling
print("\n1. Oversampling (SMOTE) - добавляем копии:")
smote = SMOTE(random_state=42)
X_over, y_over = smote.fit_resample(X, y)
print(f"   Было: {y.value_counts().to_dict()}")
print(f"   Стало: {y_over.value_counts().to_dict()}")
print("   (Классы сравнялись, добавились искусственные данные)")

# 2. Undersampling
print("\n2. Undersampling (RandomUnderSampler) - удаляем лишнее:")
undersampler = RandomUnderSampler(random_state=42)
X_under, y_under = undersampler.fit_resample(X, y)
print(f"   Было: {y.value_counts().to_dict()}")
print(f"   Стало: {y_under.value_counts().to_dict()}")
print("   (Классы сравнялись, удалили лишние данные большинства)")
```
