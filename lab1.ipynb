## 4. Группировка, сортировка и создание новых колонок

- `groupby()` — группировка данных
- `sort_values()` — сортировка
- Создание новых колонок через арифметику

```python
# Преобразуем строку Date в datetime и создадим колонку "Year"
df["Year"] = pd.to_datetime(df["Date"]).dt.year

print("Колонка Year добавлена. Первые 5 значений:")
print(df[["Date", "Year"]].head())
```

```python
# Создаём колонку "Price_Range" = разница между High и Low
df["Price_Range"] = df["High"] - df["Low"]

print("Колонка Price_Range создана. Первые 5 значений:")
print(df[["Date", "High", "Low", "Price_Range"]].head())
```

```python
# Группируем по году и считаем среднюю цену закрытия
yearly_avg = df.groupby("Year")["Close"].mean()

print("Средняя цена золота по годам:")
print(yearly_avg)
```

```python
# Сортируем по цене закрытия (по убыванию) — топ-5 самых дорогих дней
top5 = df.sort_values(by="Close", ascending=False).head(5)

print("Топ-5 самых дорогих дней по цене Close:")
print(top5[["Date", "Close", "Price_Range"]])
```

## 5. Удаление колонок и работа с пропусками

- `drop()` — удаление строк/колонок
- `isnull().sum()` — проверка пропусков
- `dropna()` — удалить строки с пропусками
- `fillna()` — заполнить пропуски

```python
# Проверим, в каких колонках есть пропуски
missing = df.isnull().sum()
missing_only = missing[missing > 0]

print("Колонки с пропусками (только те, где > 0):")
print(missing_only)
```

```python
# Удаляем ненужные колонки, оставим только ключевые для анализа
# Копируем df, чтобы не портить оригинал
df_clean = df[["Date", "Open", "High", "Low", "Close", "Volume", "Year", "Price_Range"]].copy()

print("Исходный размер:", df.shape)
print("После отбора колонок:", df_clean.shape)
print()
print(df_clean.head())
```

```python
# Демонстрация: создадим крохотный DataFrame с пропусками
demo = pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie", "Diana"],
    "Age": [25, None, 30, None],
    "City": ["Moscow", "SPb", None, "Kazan"]
})

print("Исходный mini-DataFrame:")
print(demo)
print()

# Удаляем строки с любым пропуском
print("После dropna() — удалены строки с NaN:")
print(demo.dropna())
print()

# Заполняем пропуски значением
print("После fillna(0) — NaN заменены на 0:")
print(demo.fillna(0))
print()

# Заполняем пропуски средним значением (для Age)
demo["Age"] = demo["Age"].fillna(demo["Age"].mean())
print("После fillna(среднее) для Age:")
print(demo)
```

```python
# Тот же график, но с добавлением скользящего среднего (30 дней)
plt.figure(figsize=(12, 6))
plt.plot(df_clean["Date_parsed"], df_clean["Close"], label="Цена (день)", alpha=0.4, color="gold")
plt.plot(df_clean["Date_parsed"], df_clean["Close"].rolling(30).mean(), label="Скользящее среднее (30 дней)", color="darkred", linewidth=2)
plt.title("Динамика цены золота со сглаживанием")
plt.xlabel("Дата")
plt.ylabel("Цена закрытия (USD)")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Матрица корреляции между числовыми колонками
numeric_cols = ["Open", "High", "Low", "Close", "Volume", "Price_Range"]
corr_matrix = df_clean[numeric_cols].corr()

print("Матрица корреляции:")
print(corr_matrix.round(3))

# Визуализация корреляции
plt.figure(figsize=(8, 6))
plt.imshow(corr_matrix, cmap="coolwarm", vmin=-1, vmax=1)
plt.colorbar(label="Коэффициент корреляции")
plt.xticks(range(len(numeric_cols)), numeric_cols, rotation=45)
plt.yticks(range(len(numeric_cols)), numeric_cols)
plt.title("Тепловая карта корреляции")

# Подпишем числа внутри ячеек
for i in range(len(numeric_cols)):
    for j in range(len(numeric_cols)):
        plt.text(j, i, f"{corr_matrix.iloc[i, j]:.2f}", ha="center", va="center",
                 color="white" if abs(corr_matrix.iloc[i, j]) > 0.5 else "black")

plt.tight_layout()
plt.show()
```

## 7. Выводы

По результатам анализа датасета цен на золото (2011–2018) можно сделать следующие выводы:

1. **Динамика цены.** С 2011 по 2012 год цена золота росла, достигнув пика (~173 USD за унцию в октябре 2012). Затем последовал спад до 2015 года (минимум ~105 USD). С 2016 года рынок стабилизировался в диапазоне 115–130 USD.

2. **Распределение цен.** Цена закрытия имеет форму, близкую к бимодальной — виден пик в районе 110–130 и второй — в районе 160–175. Это отражает два режима рынка: «спокойный» и «пиковый».

3. **Волатильность.** Дневной размах цены (High − Low) наиболее высок в 2013 году — это период активного падения золота. В 2016–2018 годах волатильность заметно снизилась.

4. **Корреляция.** Цены Open, High, Low, Close сильно коррелируют между собой (>0.9) — классический пример мультиколлинеарности. Для линейных моделей достаточно оставить одну из этих колонок. Объём торгов слабо связан с ценой.

5. **Признаки (features).** Созданы две новые колонки: `Year` (год) и `Price_Range` (дневной размах). Это пример *feature engineering* — создания признаков на основе существующих данных.

```python
import pandas as pd

df = pd.read_csv("data/gold.csv")
df.head()
```

```python
# 1. Создаём признак "Direction" — направление изменения цены закрытия
# Сравниваем текущую цену закрытия с предыдущей
df["Direction"] = "Down"  # по умолчанию считаем, что цена упала
df.loc[df["Close"] > df["Close"].shift(1), "Direction"] = "Up"
# Первая строка станет NaN, потому что нет предыдущего дня — удалим её
df = df.dropna(subset=["Direction"])

# 2. Считаем количество дней с ростом и падением
direction_counts = df["Direction"].value_counts()
print(direction_counts)

# 3. Строим круговую диаграмму
import matplotlib.pyplot as plt

plt.figure(figsize=(6, 6))
plt.pie(
    direction_counts,
    labels=direction_counts.index,
    autopct="%1.1f%%",      # показываем проценты на диаграмме
    colors=["#66b3ff", "#ff9999"],  # синий для Up, красный для Down
    startangle=90
)
plt.title("Распределение дней по направлению изменения цены золота")
plt.axis("equal")  # чтобы круг был ровным, а не овалом
plt.show()
```

Синяя часть (Down): 50.1% — это дни, когда цена золота падала.

Красная часть (Up): 49.9% — это дни, когда цена золота росла.

В абсолютных числах: 860 дней падения и 858 дней роста. Разница всего в 2 дня!
